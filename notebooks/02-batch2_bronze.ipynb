```python
from pyspark.sql.functions import *
```

```python
cleaned_df=spark.read.table(
    "grocery_delivery_project.bronze.clean_batch02"
)
```

```python
bronze_df02=(
    cleaned_df

       .withColumn(
        "batch_id", 
        lit("BATCH_002")
                )
    .withColumn(
        "source",
         lit("GROCERY-INCREMENTAL")
)
    .withColumn(
        "ingestion_time",
        current_timestamp()
                )
)
```

```python
display(bronze_df02)
```

```python
bronze_df02.write\
    .mode("overwrite")\
        .format("delta")\
        .option("overwriteSchema", "true")\
       .saveAsTable("grocery_delivery_project.bronze.bronze_df02")
```
