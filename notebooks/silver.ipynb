```python
from pyspark.sql.functions import *
```

```python
bronze_df=spark.read.table(
    "grocery_delivery_project.bronze.bronze_table"
)
```

```python
from pyspark.sql.functions import *
```

```python
silver_df=(
    bronze_df

    .withColumn(
        "customer_name",
        upper(trim(col("customer_name")))
    )
     .withColumn(
        "city",
        upper(trim(col("city")))
    )
      .withColumn(
        "product_name",
        upper(trim(col("product_name")))
    )
      
       .withColumn(
        "net_sales",
        round(
            (col("unit_price") * col("quantity")) - col("discount"), 2)
        )
       )
```

```python
display(silver_df)
```

```python
silver_df.write\
    .mode("overwrite")\
        .format("delta")\
            .saveAsTable("grocery_delivery_project.silver.silver_table")
```
