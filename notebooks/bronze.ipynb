```python
from pyspark.sql.functions import *
```

```python
cleaned_df=spark.read.table(
    "grocery_delivery_project.bronze.bronze_table"
)
```

```python
bronze_df=(
    cleaned_df

    .withColumn(
        "batch_id", 
        lit("BATCH_001")
                )
    .withColumn(
        "source",
         lit("GROCERY-INITIAL")
)
    .withColumn(
        "ingestion_time",
        current_timestamp()
                )
)
```

```python
display(bronze_df)
```

```python
bronze_df.write\
    .mode("overwrite")\
        .format("delta")\
            .option("mergeSchema", "true")\
                .saveAsTable("grocery_delivery_project.bronze.bronze_table")


display(
    spark.read.table(
        "grocery_delivery_project.bronze.bronze_table"
    )
)
```
